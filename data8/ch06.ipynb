```python
pokemon = Table.read_table('data/pokemon.csv')
pokemon
```

```python
summed_values = pokemon.column('against_bug') + pokemon.column('against_dark') + pokemon.column('against_dragon') + pokemon.column('against_electric') + pokemon.column('against_fairy') + pokemon.column('against_fight') + pokemon.column('against_fire') + pokemon.column('against_flying') + pokemon.column('against_ghost') + pokemon.column('against_grass') + pokemon.column('against_ground') + pokemon.column('against_ice') + pokemon.column('against_normal') + pokemon.column('against_poison') + pokemon.column('against_psychic') + pokemon.column('against_rock') + pokemon.column('against_steel') + pokemon.column('against_water')
pokemon.select('name', 'type1', 'type2', 'generation').with_column('loser rating', summed_values).sort('loser rating', descending=True).show(25)
```

```python
temp = pokemon.select('name', 'is_legendary', 'generation')
temp
```

```python
gen1 = temp.where('generation', are.equal_to(1))
gen1.num_rows
```

```python
gen1.where('is_legendary', 1).num_rows / gen1.num_rows
```

```python
which_gen = 2
gen = temp.where('generation', which_gen)
gen.where('is_legendary', 1).num_rows / gen.num_rows
```

```python
which_gen = 3
gen = temp.where('generation', which_gen)
gen.where('is_legendary', 1).num_rows / gen.num_rows
```

```python
temp = pokemon.select('name', 'height_m')
temp.show(25)
```

```python
temp.where('height_m', are.above_or_equal_to(0)).sort('height_m', descending=True).show(5)
```

```python
temp = pokemon.select('name', 'speed', 'weight_kg').with_column(
    'speed/weight ratio', pokemon.column('speed') / pokemon.column('weight_kg')
)
temp
```

```python
# nan is short for "not a number". It's a value that is a placeholder for "I don't know"
# you can use .where to get rid of nans
temp.where(3, are.above_or_equal_to(0)).sort(3, descending=True)
```

```python
# question: how does the speed of pokemon that are more likely to be female
# compare to the speed of pokemon that are more likely to be male?
temp = pokemon.select('name', 'percentage_male', 'speed')
temp
```

```python
temp = temp.with_column('percentage_female', 100 - temp.column('percentage_male'))
temp
```

```python
temp.where('percentage_female', are.above(50))
```

```python
temp.where('percentage_female', are.above(50)).column('speed').mean()
```

```python
temp.where('percentage_male', are.above(50)).column('speed').mean()
```
